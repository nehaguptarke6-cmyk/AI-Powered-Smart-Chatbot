
# AI-Powered Smart Chatbot

## Final Capstone Project

### Project Overview
This project develops an interactive AI chatbot using Python, Streamlit, LangChain, Ollama, and the Llama 3.2 3B language model.

The chatbot accepts questions from users and generates AI-based responses using a locally running language model.



## 1. Problem Statement

Users need a simple interface to ask questions and receive natural-language responses from an AI language model.

The objective is to build a chatbot that connects a web interface to a local language model and supports conversational interaction.

## 2. Objectives

- Build an interactive chatbot interface.
- Integrate a local LLM using LangChain and Ollama.
- Support questions and follow-up messages.
- Maintain conversation history during a session.
- Test the chatbot and document its limitations.



## 3. System Architecture

User Question → Streamlit Interface → LangChain Message History → Ollama → Llama 3.2 3B → AI Response → Streamlit Interface

## 4. Workflow

1. The user enters a question in the Streamlit interface.
2. The application adds the question to the conversation history.
3. LangChain sends the messages to the local model through Ollama.
4. The model generates a response.
5. The application displays and stores the response for the current session.


```python

import streamlit as st
from langchain_ollama import ChatOllama
from langchain_core.messages import (
    SystemMessage,
    HumanMessage,
    AIMessage,
)

print("Required libraries imported successfully.")
```


## 5. Model Configuration

The application uses the Llama 3.2 3B model through Ollama.

- LangChain manages messages and model interaction.
- Ollama runs the language model locally.
- Temperature is set to 0.3 to encourage relatively consistent responses.

Before running the next cell, install Ollama and download the model using:

`ollama pull llama3.2:3b`


```python

llm = ChatOllama(
    model="llama3.2:3b",
    temperature=0.3,
)

print("Chatbot model configured.")
```

```python

question = "What is machine learning?"

response = llm.invoke([
    SystemMessage(
        content="Answer clearly and use simple language."
    ),
    HumanMessage(content=question),
])

first_answer = str(response.content)

print("Question:", question)
print("\nAI Response:")
print(first_answer)
```

```python

follow_up_question = "Explain it in even simpler words."

follow_up_response = llm.invoke([
    SystemMessage(
        content="Answer clearly and use simple language."
    ),
    HumanMessage(content=question),
    AIMessage(content=first_answer),
    HumanMessage(content=follow_up_question),
])

print("Follow-up Question:", follow_up_question)
print("\nAI Response:")
print(follow_up_response.content)
```

```python

programming_question = "Explain a Python list with a simple example."

programming_response = llm.invoke([
    SystemMessage(
        content="You are a helpful Python tutor. Use simple language."
    ),
    HumanMessage(content=programming_question),
])

print("Question:", programming_question)
print("\nAI Response:")
print(programming_response.content)
```
